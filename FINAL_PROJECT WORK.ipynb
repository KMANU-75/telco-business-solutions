## TASK 1 — PREDICTING REVENUE (REGRESSION)
# Objective: Address the Finance team's request by predicting MonthlyCharges based on subscriber profiles.

```python
# IMPORTING THE IMPORTANT LIBRARIES
# Target choice: TotalCharges is the lifetime revenue a customer has generated, so it is the natural "revenue" target. 

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, accuracy_score
```

```python
# DEFINE FEATURES AND TARGET (TotalCharges & Churn excluded to prevent target leakage)

X = df_telco.drop(columns=['MonthlyCharges', 'TotalCharges', 'Churn'])
y = df_telco['MonthlyCharges']  
```

```python
# DISPLAY FEATURES & TARGET
X.head()
```

```python
y.head()
```

```python
# 3. ENCODING CATEGORICAL DATA WITH ONE-HOT-ENCODER

X_encoded = pd.get_dummies(X, drop_first=True)
```

```python
# DISPLAY ENCODED CATEGORICAL DATA(FEATUES)

X_encoded.head()
```

```python
# SPLITTING DATASET INTO TRAINING & TESTING SETS

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)
```

```python
# CREATING AN INSTANCE FOR THE REGRESSION ALGORITHM

modelReg = LinearRegression()
modelReg.fit(X_train, y_train)
```

```python
# MODEL EXALUATION

y_pred = modelReg.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
```

# TASK 3: SEGMENTING CUSTOMERS

## Objective: 
## Group customers into actionable marketing segments using K-Means Clustering

```python
# IMPORTING REQUIRED PYTHON LIBRARIES

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
# DISPLAYING DATASET

df_telco.head()
```

```python
# CREATING FEATURES(X) & ENCODING

X_cluster = df_telco.drop(columns=['Churn'])
X_encoded = pd.get_dummies(X_cluster, drop_first=True)
```

```python
# SCALING THE FEATURES(X) USING STANDARDSCALER

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)
```

```python
# CREATING AN INSTANCE FOR THE ALGORITHM / FITTING THE MODEL & TRAINING IT

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df_telco['Segment_Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
# DISPLAY SEGMENT CLUSTER

df_telco['Segment_Cluster']
```

```python
# CREATING PROFILE CLUSTERS FOR MARKETING CAMPAIGNS

cluster_profiles = df_telco.groupby('Segment_Cluster').agg({
    'tenure': 'mean',
    'MonthlyCharges': 'mean',
    'TotalCharges': 'mean',
    'Contract': lambda x: x.mode()[0],
    'InternetService': lambda x: x.mode()[0],
    'Churn': lambda x: (x == 'Yes').mean() * 100
}).rename(columns={'Churn': 'Observed_Churn_Rate_%'})
```

```python
# DISPLAY CLUSTER PROFILES

print("--- MARKETING TASK: CLUSTER PERSONAS ---")
print(cluster_profiles)
```

```python
# BUSINESS CONCLUSION FOR THE MARKETING TEAM

# Four actionable segments identified.
# The biggest opportunity is Segment 1 (new high-spenders on month-to-month) — high value 
# and high flight risk — where a contract-lock incentive protects the most revenue.
```
